# Movie Recommendation System - Analysis

Use this notebook for optional exploratory data analysis after placing `movies.csv` and `ratings.csv` in the `data/` folder.


```python
import pandas as pd
from pathlib import Path

data_dir = Path('../data')
movies = pd.read_csv(data_dir / 'movies.csv')
ratings = pd.read_csv(data_dir / 'ratings.csv')

movies.head(), ratings.head()
```

```python
print('Movies:', movies.shape)
print('Ratings:', ratings.shape)
print('Users:', ratings['userId'].nunique())
print('Average rating:', ratings['rating'].mean())
```
